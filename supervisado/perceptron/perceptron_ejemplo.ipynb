# Entrenamiento de un perceptrón para hacer deporte

## Contexto del problema

Una persona desea practicar deportes pero para tomar esa decisión necesita evaluar 2 condiciones:
- x1: Temperatura agradable (1- Si es agradable; 0- Si no es agradable)
- x2: Lluvia (1- Si no esta lloviendo; 0- Si esta lloviendo)

Las posibles (y) salidas respecto a la evaluacion de estas condiciones:
- 1: Sale a practicar deporte
- 0: Se queda en casa

El perceptron debe considerar ambas condiciones favorables para que una persona salga, por lo tanto, tendrá un comportamiento como una compuerta **AND**

```python
# Importación de bibliotecas
import numpy as np
from sklearn.linear_model import Perceptron
```

```python
#Definición de los datos de entrenamiento
X = np.array([
    [0, 0], #Clima desagradable y esta lloviendo   
    [0, 1], #Clima desagradable y no esta lloviendo
    [1, 0], #Clima agradable y esta lloviendo
    [1, 1]  #Clima agradable y no esta lloviendo
])
```

```python
#DEfinicion del vector con las posibles salidas (y)
y = np.array([0, 0, 0, 1]) #0: No salir, 1: Salir
```

## Entrenamiento del perceptron

```python
# Hiperparámetros del perceptrón
# max_iter: Número máximo de iteraciones para el entrenamiento
# tol: Tolerancia para la convergencia (si el cambio en los pesos es menor que este valor, el entrenamiento se detiene)
perceptron = Perceptron(max_iter=1000, tol=1e-1, random_state=42)
```

```python
#entrenamiento del modelo
perceptron.fit(X, y)
print("Entrenamiento completado.")
```

## Evaluación del perceptrón

```python
#Creación de un predictor
predictor = perceptron.predict(X)
print("--Resultados de las predicciones del perceptrón--")

for i in range(len(X)):
    print(f"Entrada: {X[i]} -> Predicción: {predictor[i]} (Valor real: {y[i]})")   
```

```python
# Evaluar la presición del modelo
accuracy = np.sum(predictor == y) / len(y)
print(f"Precisión del modelo: {accuracy * 100:.0f}%")
```
